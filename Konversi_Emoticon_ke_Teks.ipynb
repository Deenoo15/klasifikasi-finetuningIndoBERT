## KONVERSI EMOTICON


```python
from google.colab import files
import pandas as pd

uploaded = files.upload()
df = pd.read_csv(list(uploaded.keys())[0])
print(df.head())
print(df.shape)

import re
emoticon_map = {
    #Sedih
    "😭" : ":'(",
    "😢" : ":(",
    "😔" : ":(",
    "😞" : ":(",
    "🥺" : ":(",
    "🥹" : ":(",
    "🥲" : ":(",
    "☹"  : ":(",
    "💔" : ":(",
    "😓" : ":(",
    "😰" : ":(",
    "😖" : ":(",
    "🤧" : ":(",
    "😪" : ":(",
    "🤒" : ":(",

    #Senang
    "😊" : ":)",
    "🙂" : ":)",
    "☺"  : ":)",
    "😚" : ":)",
    "😘" : ":)",
    "😇" : ":)",
    "😌" : ":)",
    "😎" : ":)",
    "😋" : ":)",
    "😻" : ":)",
    "😸" : ":)",
    "🥰" : ":)",
    "😄" : ":D",
    "😀" : ":D",
    "😁" : ":D",
    "😆" : ":D",
    "😂" : ":'D",
    "🤣" : ":'D",
    "😅" : ":)",

    #Kaget
    "😮" : ":o",
    "😱" : ":o",
    "🤯" : ":o",
    "😵" : ":o",
    "🫠" : ":o",

    #Marah
    "😩" : ">:(",
    "😫" : ">:(",
    "😤" : ">:(",
    "😡" : ">:(",
    "🤡" : ">:(",
    "👺" : ">:(",

    #Lelah
    "😴" : "|-)",
    "🥵" : "|-)",

    #Sinis
    "🙃" : ":)",
    "😏" : ":)",
    "😑" : ":|",
    "😶" : ":|",
    "🙄" : ":|",
    "😒" : ":|",
    "😬" : ":|",
    "🤐" : ":|",

    #Bingung
    "🤔" : ":?",
    "🥴" : ":/",

    #Sering muncul
    "😮💨" : ":o",
    "ㅠㅠ" : ":(",
}

def convert_emoticon(text):
    if not isinstance(text, str):
        return text
    # Urutkan dari yang terpanjang agar tidak bentrok
    for emoticon in sorted(emoticon_map.keys(), key=len, reverse=True):
        text = text.replace(emoticon, emoticon_map[emoticon])
    # Hapus emoji/karakter non-ASCII yang tersisa yang bukan simbol penting
    text = re.sub(r'[^\x00-\x7F\u00C0-\u024F\u1E00-\u1EFF]+', '', text)
    return text.strip()

df['cleaning'] = df['cleaning'].apply(convert_emoticon)
print(df.head(10))

df.to_csv('Data_Siap_Converted.csv', index=False, sep=';', encoding='utf-8-sig')
print("Selesai!")
```
